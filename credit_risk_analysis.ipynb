# Credit Risk Analysis

## Imports and Variables

```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
CREDIT_RISK_PATH = Path("datasets/credit_risk.csv")
```

## Dataset

```python
credit_risk_df = pd.read_csv(CREDIT_RISK_PATH)
display(credit_risk_df.sample(5))
```

## Starting Analysis

```python
credit_risk_description_df = credit_risk_df.describe()
display(credit_risk_description_df)
```

```python
default, counts = np.unique(
    ar=credit_risk_df['default'],
    return_counts=True
)
credit_risk_default_count_df = pd.DataFrame({
    'default': default,
    'count': counts
})
display(credit_risk_default_count_df)
```

## Data Visualization

```python
sns.countplot(x=credit_risk_df['default'])
```

```python
plt.hist(x=credit_risk_df['age'])
```

```python
px.scatter_matrix(
    data_frame=credit_risk_df,
    dimensions=["age", "income"],
    color="default"
).show()
```
