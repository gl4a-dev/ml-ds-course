# Credit Risk Analysis with SVM

# Imports

```python
from pathlib import Path
import pickle
import numpy as np

from sklearn.metrics import classification_report, confusion_matrix
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV
```

## Load Databases

```python
PREPROCESSED_CREDIT_RISK_DATASET = Path('../datasets/credit_risk.pkl')

with open(PREPROCESSED_CREDIT_RISK_DATASET, mode="rb") as f:
    X_credit_risk_train, X_credit_risk_test, y_credit_risk_train, y_credit_risk_test = pickle.load(f)

print(X_credit_risk_train.shape)
print(y_credit_risk_train.shape)
print(X_credit_risk_test.shape)
print(y_credit_risk_test.shape)
```

## Training

```python
svm_model = SVC(
    kernel='rbf',
    C=2.0,
    random_state=1
).fit(
    X=X_credit_risk_train,
    y=y_credit_risk_train
)
```

## Test Model

```python
y_credit_risk_pred = svm_model.predict(X=X_credit_risk_test)
```

```python
cr = classification_report(
    y_true=y_credit_risk_test,
    y_pred=y_credit_risk_pred
)
print(cr)
```

```python
cm = confusion_matrix(
    y_true=y_credit_risk_test,
    y_pred=y_credit_risk_pred
)
print(cm)
```

## Tuning

### Load datasets

```python
X_credit_risk = np.concatenate((X_credit_risk_train, X_credit_risk_test), axis=0)
y_credit_risk = np.concatenate((y_credit_risk_train, y_credit_risk_test), axis=0)

print(X_credit_risk.shape)
print(y_credit_risk.shape)
```

### Grid Search

```python
params = {
    'tol': [0.001, 0.0001, 0.00001, 0.000001],
    'C': [1.0, 1.5, 2.0],
    'kernel': ['rbf', 'linear', 'poly', 'sigmoid']
}

grid_search = GridSearchCV(
    estimator=SVC(),
    param_grid=params
).fit(X_credit_risk, y_credit_risk)

print(grid_search.best_params_)
print(grid_search.best_score_)
```
