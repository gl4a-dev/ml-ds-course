# Credit Risk Analysis with MLP

# Imports

```python
from pathlib import Path
import pickle
import numpy as np

from sklearn.metrics import classification_report, confusion_matrix
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import GridSearchCV
```

## Load Databases

```python
PREPROCESSED_CREDIT_RISK_DATASET = Path('../datasets/credit_risk.pkl')

with open(PREPROCESSED_CREDIT_RISK_DATASET, mode="rb") as f:
    X_credit_risk_train, X_credit_risk_test, y_credit_risk_train, y_credit_risk_test = pickle.load(f)

print(X_credit_risk_train.shape)
print(y_credit_risk_train.shape)
print(X_credit_risk_test.shape)
print(y_credit_risk_test.shape)
```

## Training

```python
mlp_model = MLPClassifier(
    hidden_layer_sizes=(2, 2),
    solver='adam',
    activation='relu',
    max_iter=1500,
    tol=10e-9,
    verbose=True
).fit(
    X=X_credit_risk_train,
    y=y_credit_risk_train
)
```

## Test Model

```python
y_credir_risk_pred = mlp_model.predict(X=X_credit_risk_test)
```

```python
cr = classification_report(
    y_pred=y_credir_risk_pred,
    y_true=y_credit_risk_test
)
print(cr)
```

```python
cm = confusion_matrix(
    y_pred=y_credir_risk_pred,
    y_true=y_credit_risk_test
)
print(cm)
```

### Tuning

### Load datasets

```python
X_credit_risk = np.concatenate((X_credit_risk_train, X_credit_risk_test), axis=0)
y_credit_risk = np.concatenate((y_credit_risk_train, y_credit_risk_test), axis=0)

print(X_credit_risk.shape)
print(y_credit_risk.shape)
```

### Grid Search

```python
params = {
    'hidden_layer_sizes': [(2, 2)],
    'activation': ['relu', 'logistic', 'tanh'],
    'solver': ['adam', 'sgd'],
    'batch_size': [10, 56]
}

grid_search = GridSearchCV(
    estimator=MLPClassifier(),
    param_grid=params
).fit(X_credit_risk, y_credit_risk)

print(grid_search.best_params_)
print(grid_search.best_score_)
```
