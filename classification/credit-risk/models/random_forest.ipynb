# Credit Risk Analysis with Random Forest

# Imports

```python
import pickle
import numpy as np
from pathlib import Path

from sklearn.metrics import classification_report, confusion_matrix
from sklearn.ensemble import RandomForestClassifier

from sklearn.model_selection import GridSearchCV
```

## Load Databases

```python
PREPROCESSED_CREDIT_RISK_DATASET = Path('../datasets/credit_risk.pkl')

with open(PREPROCESSED_CREDIT_RISK_DATASET, mode="rb") as f:
    X_credit_risk_train, X_credit_risk_test, y_credit_risk_train, y_credit_risk_test = pickle.load(f)

print(X_credit_risk_train.shape)
print(y_credit_risk_train.shape)
print(X_credit_risk_test.shape)
print(y_credit_risk_test.shape)
```

## Training

```python
random_forest_model = RandomForestClassifier(
    n_estimators=40,
    criterion='entropy',
    random_state=0
).fit(
    X=X_credit_risk_train,
    y=y_credit_risk_train
)
```

# Test Model

```python
y_credit_risk_pred = random_forest_model.predict(X=X_credit_risk_test)
```

```python
cr = classification_report(
    y_true=y_credit_risk_test,
    y_pred=y_credit_risk_pred
)
print(cr)
```

```python
cm = confusion_matrix(
    y_true=y_credit_risk_test,
    y_pred=y_credit_risk_pred
)
print(cm)
```

## Tuning

### Dataset for cross validation

```python
X_credit_risk = np.concatenate(
    (X_credit_risk_train, X_credit_risk_test),
    axis=0
)
y_credit_risk = np.concatenate(
    (y_credit_risk_train, y_credit_risk_test),
    axis=0
)

print(X_credit_risk.shape)
print(y_credit_risk.shape)
```

### Grid Search

```python
params = {
    'criterion': ['gini', 'entropy'],
    'n_estimators': [10, 40, 50, 60, 10, 150],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 5, 10]
}

grid_serach = GridSearchCV(
    estimator=RandomForestClassifier(),
    param_grid=params
).fit(
    X=X_credit_risk, 
    y=y_credit_risk
)

print(grid_serach.best_params_)
print(grid_serach.best_score_)
```
