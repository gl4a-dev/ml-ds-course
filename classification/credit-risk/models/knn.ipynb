# Credit Risk Analysis with KNN

# Imports

```python
from pathlib import Path
import pickle
import numpy as np

from sklearn.metrics import classification_report, confusion_matrix
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import GridSearchCV
```

## Load Databases

```python
PREPROCESSED_CREDIT_RISK_DATASET = Path('../datasets/credit_risk.pkl')

with open(PREPROCESSED_CREDIT_RISK_DATASET, mode="rb") as f:
    X_credit_risk_train, X_credit_risk_test, y_credit_risk_train, y_credit_risk_test = pickle.load(f)

print(X_credit_risk_train.shape)
print(y_credit_risk_train.shape)
print(X_credit_risk_test.shape)
print(y_credit_risk_test.shape)
```

## Training

```python
knn_model = KNeighborsClassifier(
    n_neighbors=5,
    metric="minkowski",
    p=2
).fit(
    X=X_credit_risk_train,
    y=y_credit_risk_train
)
```

## Test Model

```python
y_credit_risk_pred = knn_model.predict(X=X_credit_risk_test)
```

```python
cr = classification_report(
    y_true=y_credit_risk_test,
    y_pred=y_credit_risk_pred
)
print(cr)
```

```python
cm = confusion_matrix(
    y_true=y_credit_risk_test,
    y_pred=y_credit_risk_pred
)
print(cm)
```

## Tuning

### Load Datasets

```python
X_credit_risk = np.concatenate((X_credit_risk_train, X_credit_risk_test), axis=0)
y_credit_risk = np.concatenate((y_credit_risk_train, y_credit_risk_test), axis=0)

print(X_credit_risk.shape)
print(y_credit_risk.shape)
```

### Grid Search

```python
params = {
    'n_neighbors': [3, 5, 10, 20],
    'p': [1, 2]
}

grid_search = GridSearchCV(
    estimator=KNeighborsClassifier(),
    param_grid=params
).fit(
    X=X_credit_risk,
    y=y_credit_risk
)

print(grid_search.best_params_)
print(grid_search.best_score_)
```
