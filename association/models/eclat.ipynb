# Market Basket Items Association with ECLAT

```python
import pandas as pd
from pyECLAT import ECLAT
```

## Dataset

### Import and Conversion

```python
market_baskets_df = pd.read_csv("../datasets/market-basket.csv", header=None)
display(market_baskets_df)
```

## Rules induction

### Fit

```python
eclat = ECLAT(data=market_baskets_df)
indexes, support = eclat.fit(
    min_support=0.01,
    min_combination=1,
    max_combination=3
)

print(support)
```

### Format

```python
rules_list = [{"ITEMS": field, "SUPPORT": value} for field, value in support.items()]
rules_df = pd.DataFrame(rules_list).sort_values(
    by="SUPPORT",
    ascending=False
).reset_index(drop=True)

display(rules_df)
rules_df.to_csv("eclat_rules.csv", index=False)
```
