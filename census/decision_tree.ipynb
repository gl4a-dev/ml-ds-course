# Census Analysis with Decision Trees

## Imports

```python
from pathlib import Path
import pickle

import matplotlib.pyplot as plt

from sklearn.metrics import classification_report, confusion_matrix
from sklearn.tree import DecisionTreeClassifier

from sklearn import tree
```

## Load Datasets

```python
PREPROCESSED_CENSUS_DATASET_PATH = Path('datasets/census.pkl')

with open(PREPROCESSED_CENSUS_DATASET_PATH, mode="rb") as f:
    X_census_train, X_census_test, y_census_train, y_census_test = pickle.load(f)

print(X_census_train.shape)
print(y_census_train.shape)
print(X_census_test.shape)
print(y_census_test.shape)
```

## Training

```python
decision_tree_model = DecisionTreeClassifier(criterion='entropy', ).fit(
    X=X_census_train, 
    y=y_census_train
)
```

## Results

### Tree

```python
fig, axes = plt.subplots(figsize=(20, 15))
tree.plot_tree(
    decision_tree=decision_tree_model,
    filled=True
); # Too long to generate, too hard to see
```

### Metrics

```python
y_census_pred = decision_tree_model.predict(X=X_census_test)
```

```python
cr = classification_report(
    y_pred=y_census_pred,
    y_true=y_census_test
)
print(cr)
```

```python
cm = confusion_matrix(
    y_pred=y_census_pred,
    y_true=y_census_test
)
print(cm)
```
