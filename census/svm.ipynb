# Census Analysis with SVM

## Imports

```python
from pathlib import Path
import pickle

from sklearn.metrics import classification_report, confusion_matrix
from sklearn.svm import SVC
```

## Load Datasets

```python
PREPROCESSED_CENSUS_DATASET_PATH = Path('datasets/census.pkl')

with open(PREPROCESSED_CENSUS_DATASET_PATH, mode="rb") as f:
    X_census_train, X_census_test, y_census_train, y_census_test = pickle.load(f)

print(X_census_train.shape)
print(y_census_train.shape)
print(X_census_test.shape)
print(y_census_test.shape)
```

## Training

```python
svm_model = SVC(
    kernel='linear',
    C=1.0,
    random_state=1
).fit(
    X=X_census_train,
    y=y_census_train
)
```

## Evaluation

```python
y_census_pred = svm_model.predict(X=X_census_test)
```

```python
cr = classification_report(
    y_pred=y_census_pred,
    y_true=y_census_test
)
print(cr)
```

```python
cm = confusion_matrix(
    y_pred=y_census_pred,
    y_true=y_census_test
)
print(cm)
```
