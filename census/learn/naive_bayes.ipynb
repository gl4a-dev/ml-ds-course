# Census Analysis with Naive Bayes

## Imports

```python
from pathlib import Path
import pickle

from sklearn.metrics import classification_report, confusion_matrix
from sklearn.naive_bayes import GaussianNB
```

## Load dataset

```python
PREPROCESSED_CENSUS_DATASET_PATH = Path('datasets/census.pkl')

with open(PREPROCESSED_CENSUS_DATASET_PATH, mode="rb") as f:
    X_census_train, X_census_test, y_census_train, y_census_test = pickle.load(f)

print(X_census_train.shape)
print(X_census_test.shape)
print(y_census_train.shape)
print(y_census_test.shape)
```

## Algorithm Training

```python
naive_bayes_model = GaussianNB().fit(
    X=X_census_train,
    y=y_census_train
)
```

## Algorithm Evaluations

```python
y_census_predict = naive_bayes_model.predict(X=X_census_test)
```

```python
cr = classification_report(
    y_true=y_census_test,
    y_pred=y_census_predict
)
print(cr)
```

```python
cm = confusion_matrix(
    y_true=y_census_test,
    y_pred=y_census_predict
)
print(cm)
```
