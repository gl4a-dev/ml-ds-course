# Credit Card Clients dataset Pre Processing

```python
import pandas as pd
```

## Pre Processing

### Load Dataset

```python
df = pd.read_csv("credit_card_clients.csv")
display(df)
display(df.describe())
```

### Aggregate Columns

```python
bill_cols = [f'BILL_AMT{i}' for i in range(1, 7)]
df['AVG_BILL_AMT'] = df[bill_cols].mean(axis=1)
df['MAX_BILL_AMT'] = df[bill_cols].max(axis=1)

pay_amt_cols = [f'PAY_AMT{i}' for i in range(1, 7)]
df['AVG_PAY_AMT'] = df[pay_amt_cols].mean(axis=1)
df['TOTAL_PAY_AMT'] = df[pay_amt_cols].sum(axis=1)

pay_status_cols = ['PAY_0', 'PAY_2', 'PAY_3', 'PAY_4', 'PAY_5', 'PAY_6']
df['MAX_DELAY'] = df[pay_status_cols].max(axis=1)
df['TOTAL_DELAYS_COUNT'] = (df[pay_status_cols] > 0).sum(axis=1)

display(df)
```

### Drop Columns

```python
df = df.drop(columns= ["ID", "default payment next month"] + bill_cols + pay_amt_cols + pay_status_cols)
display(df)
```

## Save dataset

```python
df.to_csv("credit_card_clients_p.csv", index=False)
```
