# Credit Card Clients Analysis with K-Means

```python
import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go
```

## Dataset

### Load

```python
credit_card_clients_df = pd.read_csv("../datasets/credit_card_clients_p.csv")
display(credit_card_clients_df)
```

### Scale

```python
scaler = StandardScaler()
credit_card_clients = scaler.fit_transform(credit_card_clients_df)

display(credit_card_clients)
```

## Evaluation

### Elbow Method

```python
def get_elbow_mathod_data_for_kmeans(data:np.ndarray, min_clusters:int, max_clusters:int) -> pd.DataFrame:
    results:list[dict] = []

    for i in range(min_clusters, max_clusters + 1):
        model = KMeans(n_clusters=i, random_state=0)
        model.fit(data)

        results.append({
            "CLUSTERS": i,
            "WCSS": model.inertia_
        })

    return pd.DataFrame(results)

elbow_method_df = get_elbow_mathod_data_for_kmeans(
    data=credit_card_clients,
    min_clusters=2,
    max_clusters=10
)

px.line(
    x=elbow_method_df["CLUSTERS"], 
    y=elbow_method_df["WCSS"],
    title="CLUSTER x WCSS for K-Means"
) # Use 4 clusters
```
