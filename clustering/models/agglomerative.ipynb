# Credit Card Clients Analysis with Agglomerative Clustering

```python
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import AgglomerativeClustering

from scipy.cluster.hierarchy import dendrogram, linkage
import plotly.express as px
```

## Dataset

### Load

```python
credit_card_clients_df = pd.read_csv("../datasets/credit_card_clients_p.csv")
display(credit_card_clients_df)
```

### Scale

```python
scaler = StandardScaler()
credit_card_clients = scaler.fit_transform(credit_card_clients_df)

display(credit_card_clients)
```

## Dendogram

```python
dendrogram(linkage(credit_card_clients, method="ward"));
```

## Apply Agglomerative Clustering

```python
model = AgglomerativeClustering(
    n_clusters=5,
    linkage="ward",
    metric="euclidean"
)

labels = model.fit_predict(X=credit_card_clients)
```

## PCA Scatter Plot

```python
pca = PCA(n_components=2)
instances_pca = pca.fit_transform(credit_card_clients)

px.scatter(
    x=instances_pca[:, 0],
    y=instances_pca[:, 1],
    color=model.labels_,
).show()
```
