# Credit Card Clients Analysis with Density-Based SCAN

```python
import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import DBSCAN

import plotly.express as px
```

## Dataset

### Load

```python
credit_card_clients_df = pd.read_csv("../datasets/credit_card_clients_p.csv")
display(credit_card_clients_df)
```

### Scale

```python
scaler = StandardScaler()
credit_card_clients = scaler.fit_transform(credit_card_clients_df)

display(credit_card_clients)
```

## Apply DBSCAN

```python
model = DBSCAN(
    eps=2,
    min_samples=20
).fit(X=credit_card_clients)

counts = np.unique(model.labels_, return_counts=True)
counts_df = pd.DataFrame({
    "CLUSTER": counts[0],
    "COUNT": counts[1]
})
display(counts_df)
```

## PCA Scatter Plot

```python
pca = PCA(n_components=2)
instances_pca = pca.fit_transform(credit_card_clients)

px.scatter(
    x=instances_pca[:, 0],
    y=instances_pca[:, 1],
    color=model.labels_,
).show()
```
