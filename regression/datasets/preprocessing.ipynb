# Preprocessing House Prices Dataset

```python
import pandas as pd
import pickle

from sklearn.model_selection import train_test_split
```

## Load Dataset

```python
house_prices_df = pd.read_csv("house_prices.csv")

display(house_prices_df)
display(house_prices_df.describe())
display(house_prices_df.isnull().sum())
```

## Splits

### Feature and Target

```python
X_house_prices = house_prices_df.iloc[:, 3:19].values
y_house_prices = house_prices_df.iloc[:, 2].values

print(X_house_prices)
print(y_house_prices)
```

### Train and Test

```python
X_house_prices_train, X_house_prices_test, y_house_prices_train, y_house_prices_test = train_test_split(X_house_prices, y_house_prices, test_size=0.3, random_state=1)

print(X_house_prices_train.shape)
print(X_house_prices_test.shape)
print(y_house_prices_train.shape)
print(y_house_prices_test.shape)
```

## Save the datasets

```python
with open("house_prices.pkl", mode="wb") as file:
    pickle.dump(
        [
            X_house_prices_train,
            X_house_prices_test,
            y_house_prices_train,
            y_house_prices_test
        ],
        file
    )


with open("house_prices.pkl", mode="rb") as f:
    X_house_prices_train, X_house_prices_test, y_house_prices_train, y_house_prices_test = pickle.load(f)

print(X_house_prices_train.shape)
print(X_house_prices_test.shape)
print(y_house_prices_train.shape)
print(y_house_prices_test.shape)
```
