# Evaluate Models with Hypothesis Test

```python
import pandas as pd
from scipy.stats import shapiro, f_oneway

import seaborn as sns
```

## Load Data

```python
df = pd.read_csv("results.csv")
display(df.head())
```

## Normality Test

### Histograms

```python
def plot_histogram(df: pd.DataFrame) -> None:
    df_long = df.melt(var_name="Model", value_name="Accuracy")

    chart = sns.displot(
        data=df_long, 
        x="Accuracy",
        hue="Model",
        kind="kde",
        height=6,
        aspect=1.5
    )

    chart.figure.suptitle("Accuracy Distribution of the Models")

plot_histogram(df=df)
```

### Shapiro Test

```python
def run_shapiro_test(df:pd.DataFrame, alpha:float=0.01) -> pd.DataFrame:
    results = []
    for col in df.columns:
        stat, p_value = shapiro(x=df[col])
        results.append(
            {
                'algorithm': col, 
                'stat': stat, 
                'p_value': p_value, 
                'is_normal': p_value > alpha
            }
        )

    return pd.DataFrame(results)

results = run_shapiro_test(df=df)
display(results)
```

## Hypothesis Test

### ANOVA

```python
def run_anova(df:pd.DataFrame, alpha:float = 0.01):
    _, p_value = f_oneway(*[df[col] for col in df.columns])
    if p_value > alpha:
        print(f"Non distinct set: p_value = {p_value}")
    else:
        print(f"Distinct sets: p_value = {p_value}")

run_anova(df=df, alpha=0.05)
```
