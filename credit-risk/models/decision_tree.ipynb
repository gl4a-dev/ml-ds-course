# Credit Risk Analysis with Decision Trees

# Imports

```python
import pickle
import numpy as np 
from pathlib import Path

import matplotlib.pyplot as plt

from sklearn.metrics import classification_report, confusion_matrix
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV

from sklearn import tree
```

## Load Databases

```python
PREPROCESSED_CREDIT_RISK_DATASET = Path('../datasets/credit_risk.pkl')

with open(PREPROCESSED_CREDIT_RISK_DATASET, mode="rb") as f:
    X_credit_risk_train, X_credit_risk_test, y_credit_risk_train, y_credit_risk_test = pickle.load(f)

print(X_credit_risk_train.shape)
print(y_credit_risk_train.shape)
print(X_credit_risk_test.shape)
print(y_credit_risk_test.shape)
```

# Training

```python
decision_tree_model = DecisionTreeClassifier(criterion='entropy').fit(
    X=X_credit_risk_train, 
    y=y_credit_risk_train
)
```

## Results

### Tree

```python
features = ['história', 'dívida', 'garantias', 'renda']
classes = ['0', '1']

fig, axes = plt.subplots(figsize=(20, 15))
tree.plot_tree(
    decision_tree=decision_tree_model,
    feature_names=features,
    class_names=classes,
    filled=True
);
```

### Metrics

```python
y_credit_risk_pred = decision_tree_model.predict(X=X_credit_risk_test)
```

```python
cr = classification_report(
    y_pred=y_credit_risk_pred,
    y_true=y_credit_risk_test
)
print(cr)
```

```python
cm = confusion_matrix(
    y_pred=y_credit_risk_pred,
    y_true=y_credit_risk_test
)
print(cm)
```

## Tuning

### Dataset for cross validation

```python
X_credit_risk = np.concatenate(
    (X_credit_risk_train, X_credit_risk_test),
    axis=0
)
y_credit_risk = np.concatenate(
    (y_credit_risk_train, y_credit_risk_test),
    axis=0
)

print(X_credit_risk.shape)
print(y_credit_risk.shape)
```

### Grid Search

```python
params = {
    'criterion': ['gini', 'entropy'],
    'splitter': ['best', 'random'],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 5, 10]
}

grid_serach = GridSearchCV(
    estimator=DecisionTreeClassifier(),
    param_grid=params
)
grid_serach.fit(X=X_credit_risk, y=y_credit_risk)

print(grid_serach.best_params_)
print(grid_serach.best_score_)
```
