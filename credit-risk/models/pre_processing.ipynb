# Credit Risk Analysis

## Imports and Variables

```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

import pickle
```

## Dataset

```python
CREDIT_RISK_PATH = Path("datasets/credit_risk.csv")

credit_risk_df = pd.read_csv(CREDIT_RISK_PATH)
display(credit_risk_df.sample(5, random_state=0))
```

## Starting Analysis

```python
credit_risk_description_df = credit_risk_df.describe()
display(credit_risk_description_df)
```

```python
default, counts = np.unique(
    ar=credit_risk_df['default'],
    return_counts=True
)
credit_risk_default_count_df = pd.DataFrame({
    'default': default,
    'count': counts
})
display(credit_risk_default_count_df)
```

## Data Visualization

```python
sns.countplot(x=credit_risk_df['default'])
```

```python
plt.hist(x=credit_risk_df['age'])
```

```python
px.scatter_matrix(
    data_frame=credit_risk_df,
    dimensions=["age", "income"],
    color="default"
).show()
```

## Data Treatment

### Inconsistent Values

```python
display(credit_risk_df.loc[credit_risk_df['age'] < 0])
```

```python
age_mean = (credit_risk_df[credit_risk_df['age'] > 0])['age'].mean() # 40.92770044906149
credit_risk_df.loc[credit_risk_df['age'] < 0, 'age'] = age_mean

credit_risk_df.describe()
```

### Missing Values

```python
display(credit_risk_df.isnull().sum())
```

```python
display(credit_risk_df.loc[pd.isnull(credit_risk_df['age'])])
```

```python
age_mean = credit_risk_df['age'].mean()
credit_risk_df.fillna(age_mean, inplace=True)

display(credit_risk_df.isnull().sum())
```

## Feature and Target split

```python
X_credit_risk = credit_risk_df.iloc[:, 1:4].values
y_credit_risk = credit_risk_df.iloc[:, 4].values

display(X_credit_risk)
display(y_credit_risk)
```

## Value Standardisation

```python
X_credit_risk = StandardScaler().fit_transform(X_credit_risk)
display(X_credit_risk)
```

## Train/Test Split

```python
X_credit_risk_train, X_credit_risk_test, y_credit_risk_train, y_credit_risk_test = train_test_split(
    X_credit_risk, y_credit_risk,
    test_size=0.25,
    random_state=0
)

print(X_credit_risk_train.shape)
print(X_credit_risk_test.shape)
print(y_credit_risk_train.shape)
print(y_credit_risk_test.shape)
```

## Save data

```python
CREDIT_RISK_PREPROCESSED_PATH = Path("datasets/credit_risk.pkl")

with open(CREDIT_RISK_PREPROCESSED_PATH, mode='wb') as file:
    pickle.dump(
        [
            X_credit_risk_train,
            X_credit_risk_test,
            y_credit_risk_train,
            y_credit_risk_test
        ],
        file
    )
```
