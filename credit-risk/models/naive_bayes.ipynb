# Credit Risk Analysis with Naive Bayes

## Imports

```python
from pathlib import Path
import pickle

from sklearn.metrics import classification_report, confusion_matrix
from sklearn.naive_bayes import GaussianNB
```

## Load datasets

```python
PREPROCESSED_CREDIT_RISK_DATASET = Path('datasets/credit_risk.pkl')

with open(PREPROCESSED_CREDIT_RISK_DATASET, mode="rb") as f:
    X_credit_risk_train, X_credit_risk_test, y_credit_risk_train, y_credit_risk_test = pickle.load(f)

print(X_credit_risk_train.shape)
print(X_credit_risk_test.shape)
print(y_credit_risk_train.shape)
print(y_credit_risk_test.shape)
```

## Algorithm Training

```python
naive_bayes_model = GaussianNB().fit(X=X_credit_risk_train, y=y_credit_risk_train)
```

## Algorithm Evaluation

```python
y_credit_risk_pred = naive_bayes_model.predict(X=X_credit_risk_test)
print(y_credit_risk_pred.shape)
```

```python
cr = classification_report(
    y_true=y_credit_risk_test, 
    y_pred=y_credit_risk_pred
)
print(cr)
```

```python
cm = confusion_matrix(
    y_true=y_credit_risk_test,
    y_pred=y_credit_risk_pred
)
print(cm)
```
